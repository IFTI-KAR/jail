## TF-IDF

**TF-IDF (Term Frequency-Inverse Document Frequency)** is a statistical measure that evaluates the importance of a word in a document relative to a collection (corpus). It has two main components:

*   TF (Term Frequency): Measures how often a word appears in a document.
*   IDF (Inverse Document Frequency): Reduces the impact of common words across many documents.

# Neural Network Using Tensorflow Keras

Imports

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

Loading the Spam Ham dataset

```python
# read dataset from a Google Drive File
file_link = 'https://drive.google.com/file/d/1hWX7aB-Cx4OeOsjHTE-Iq1SfxFUpf71t/view?usp=sharing' # the file access must have to be Public
id = file_link.split("/")[-2]
print(id)
new_link = f'https://drive.google.com/uc?id={id}'
print(new_link)
df = pd.read_csv(new_link)
```

```python
df.head()
```

```python
df.shape
```

Prepare labels & splits

```python
le = LabelEncoder()
y = le.fit_transform(df["Category"])  # spam=1, ham=0
X_text = df["Message"].astype(str).values

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
vec = TfidfVectorizer(max_features=1000, ngram_range=(1,2), stop_words="english")
X_train = vec.fit_transform(X_train_text)
X_test  = vec.transform(X_test_text)
print(type(X_train), 'shape = ',X_train.shape)
# Tensorflow Keras Dense expects dense arrays; convert
X_train = X_train.astype("float32").toarray()
X_test  = X_test.astype("float32").toarray()
print(type(X_train), 'shape = ',X_train.shape)
```

# Designing the model

image.png

```python
X_train.shape
```

## Building the model

```python
model = Sequential([
    Input(shape= (X_train.shape[1],) ),           # Input layer
    Dense(32, activation="relu"),                 # First Hidden Layer
    Dense(16, activation="leaky_relu"),           # Second Hidden Layer
    Dense(1, activation="sigmoid")                # Output layer
])
```

Possible Activation Functions: 'relu'
'sigmoid'
'softmax'
'tanh'
'linear'
'leaky_relu' and many more. Check Keras Documentation

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

Possible Optimizer = 'adam'
'sgd'
'rmsprop'
'adagrad'
'adadelta'

## Training the model

```python
history = model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)
```

## Evaluating the model

```python
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Loss: {loss:.4f} | Test Accuracy: {acc:.4f}")
```

```python
y_prob = model.predict(X_test, verbose=0).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le.classes_))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Accuracy (sklearn):", accuracy_score(y_test, y_pred))
```

## Plotting the loss and accuracy curve

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(history.history['loss'], label='Train Loss', marker = 'x')
plt.plot(history.history['val_loss'], label='Validation Loss', marker = 'x')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
```

# Using EarlyStopping to stop training when model starts overfitting

## Building the model

```python
model = Sequential([
    Input(shape= (X_train.shape[1],)),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dropout(0.2),
    Dense(8, activation="relu"),
    Dense(1, activation="sigmoid")  #output layer
])
```

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

## Training the model

```python
es = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)
history = model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=64,
    validation_split=0.2,
    callbacks=[es],
    verbose=1
)
```

## Evaluating the model

```python
y_prob = model.predict(X_test, verbose=0).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le.classes_))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Accuracy (sklearn):", accuracy_score(y_test, y_pred))
```

```python
plt.plot(history.history['loss'], label='Train Loss', marker = 'x')
plt.plot(history.history['val_loss'], label='Validation Loss', marker = 'x')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
```

# Multi Class Classification

Dataset: Coronavirus tweets NLP - Text Classification
[Kaggle Link](https://www.kaggle.com/datasets/datatattle/covid-19-nlp-text-classification?select=Corona_NLP_train.csv)

```python
# read dataset from a Google Drive File
file_link = 'https://drive.google.com/file/d/1girV6yooTfkf2_wrKKXTnJEGC2_uBg3L/view?usp=sharing' # the file access must have to be Public
id = file_link.split("/")[-2]
print(id)
new_link = f'https://drive.google.com/uc?id={id}'
print(new_link)
df = pd.read_csv(new_link)
```

```python
df.head()
```

```python
df.shape
```

```python
df.value_counts("Sentiment")
```

### Removing unnecessary columns

```python
df = df.drop(["UserName","ScreenName","Location","TweetAt"], axis=1)
df.head()
```

## Mapping the labels of multi class classification

```python
label_map = {
    "Extremely Negative": 0,
    "Negative": 1,
    "Neutral": 2,
    "Positive": 3,
    "Extremely Positive": 4
}
df["Sentiment"] = df["Sentiment"].map(label_map)
df.head()
```

```python
X_text = df["OriginalTweet"].astype(str).values
y = df["Sentiment"].values
```

# One hot Encoding

```python
from tensorflow.keras.utils import to_categorical
y_onehot = to_categorical(y, num_classes=5)

print(y_onehot[:5])
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y_onehot,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    stop_words="english"
)

X_train = tfidf.fit_transform(X_train_text)
X_test = tfidf.transform(X_test_text)

X_train = X_train.toarray().astype("float32")
X_test = X_test.toarray().astype("float32")

print(X_train.shape)
print(y_train.shape)
```

## Building the model

```python
model = Sequential([
    Input(shape= (X_train.shape[1],)),
    Dense(256, activation="relu"),
    Dropout(0.3),
    Dense(128, activation="relu"),
    Dropout(0.3),
    Dense(5, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
model.summary()
```

## Training the model

```python
es = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)
history = model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=64,
    verbose=1,
    callbacks=[es]
)
```

## Evaluating the model

```python
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
y_pred_probs = model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = np.argmax(y_test, axis=1)

target_names = [
    "Extremely Negative",
    "Negative",
    "Neutral",
    "Positive",
    "Extremely Positive"
]

print(classification_report(y_true, y_pred, target_names=target_names))

print(confusion_matrix(y_true, y_pred))
```

```python
plt.plot(history.history['loss'], label='Train Loss', marker = 'x')
plt.plot(history.history['val_loss'], label='Validation Loss', marker = 'x')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
```

```python
plt.plot(history.history['accuracy'], label='Train Accuracy', marker = 'x')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', marker = 'x')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
```
